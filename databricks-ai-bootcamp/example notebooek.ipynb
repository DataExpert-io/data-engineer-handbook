```python
dataframe = spark.sql("SELECT * FROM dataexpert_student.student_table")

spark.sql("DROP TABLE IF EXISTS dataexpert_student.company_aggregate")

create_table =  spark.sql("""CREATE TABLE IF NOT EXISTS dataexpert_student.company_aggregate(
    company VARCHAR(32),
    max_age INT,
    min_age INT
)""")

aggregated_dataframe = spark.sql("""SELECT 
                                 company,
                                 MAX(age) as max_age,
                                 MIN(age) as min_age 
                                 FROM dataexpert_student.student_table 
                                 GROUP BY company""")
```
