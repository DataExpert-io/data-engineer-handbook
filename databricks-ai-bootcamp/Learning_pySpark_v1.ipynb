```python
spark.sql("DROP TABLE IF EXISTS learnDB.class_aggregate")

Create_table = spark.sql(
    "CREATE TABLE IF NOT EXISTS learnDB.class_aggregate (class VARCHAR(5),Maxage INT,Minage INT, Avgage FLOAT ,TotalStudents INT, Avgmarks FLOAT)"
)

spark.sql(
    "insert into learnDB.class_aggregate select class, max(age) Maxage, min(age) Minage, avg(age) Avgage,Count(*) TotalStudents,round(Avg(m.marks),2) Avgmarks from learnDB.student s left join learnDB.marks m on s.StudentId = m.StudentId Group By class order by Maxage"
)

Dataframe = spark.sql("SELECT * FROM learnDB.class_aggregate")

Dataframe.collect()
```

```python

spark.sql("Drop table if exists learnDB.mark_aggregate")

Create_table = spark.sql(
    "CREATE TABLE IF NOT EXISTS learnDB.mark_aggregate (class VARCHAR(5),subject VARCHAR(10),avgmarks FLOAT)"
)

spark.sql(
    "insert into learnDB.mark_aggregate select class,subject,avg(marks) avgmarks from learnDB.marks left join learnDB.student on marks.StudentId = student.StudentId group by class,subject order by class,subject"
)

Dataframe1 = spark.sql("SELECT * FROM learnDB.mark_aggregate")

Dataframe1.collect()
```
